```python
# Regresión lineal simple

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm

plt.rcParams['figure.figsize'] = 20,5
```

```python
url = "https://raw.githubusercontent.com/justmarkham/scikit-learn-videos/master/data/Advertising.csv"
df = pd.read_csv(url, usecols = range(1,5))
```

```python
df.info()
```

```python
df.head()
```

```python
fig, axs = plt.subplots(1,3)
for i in range(3):
  axs[i].scatter(x = df.iloc[:,i], y = df.Sales)
  axs[i].set_xlabel(df.columns[i])
  axs[i].set_ylabel("Sales")

plt.show()
```

```python
# veamos una regresión simple
```

```python
# Importamos el módulo

from sklearn.linear_model import LinearRegression
```

```python
# Instanciamos la clase de la regresion lineal

model = LinearRegression(fit_intercept= True)
```

```python
# Definimos los datos para el modelo

X = df[["TV"]] #mantiene la estructura bidimensional
y = df["Sales"]
```

```python
# Entrenemos el modelo

model.fit(X,y)
```

```python
# veamos los atributos del modelo
model.coef_ , model.intercept_
```

```python
# con los parametros calculados podemos predecir

model.predict([[4]])
```
