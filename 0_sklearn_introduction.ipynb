```python
# sckit-learn provides dozens of buil-in algorithmns and models calles estimators
# each estimartos can be fitted to some data using its fit method

from sklearn.ensemble import RandomForestClassifier

clf = RandomForestClassifier(random_state = 0)
X = [[1,2,3],
     [11,12,13]]
y = [0,1]
clf.fit(X,y)

# Fit accepts 2 inputs: sample matrix X and targetn values y (1d array), both are usually numpy arrays or equivalent
```

```python
# Onece tue estimator is fitted, it can be used for predicting target values of new data

clf.predict(X) # predict classes of the training data
```

```python
clf.predict([[4,5,6],
            [14,15,16]]) # predict classes of new data
```

```python
# ML workflows are often composed of different parts
# A ypical pipeline consists of 1. a pre-processing step that transforms or imputes the data
# 2. a final predictor that predicts target values

from sklearn.preprocessing import StandardScaler
X = [[0,15],
     [1,-10]]

# scale data according to computed scaling values

# The transformer objects don't hava a predict method but rather a trasnform method

StandardScaler().fit(X).transform(X)
```

```python
# Transformers and estimators (predictors) can be combined together into a single unifying object: a Pipeline

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Create a pipeline object

pipe = make_pipeline(
    StandardScaler(),
    LogisticRegression()
)

# Load the iris dataset and split it into train and tests sets
X, y = load_iris(return_X_y = True)
X_train, X_test, y_train, y_test = train_test_split(X,y,random_state = 0)

# fit the wholes pipeline

pipe.fit(X_train, y_train)

# we can now use it like any other estimator

accuracy_score(pipe.predict(X_test), y_test)
```

```python
# Model evaluation

# Let's see how to perform a 5-fold-cross-validation procedure

from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_validate

# Creating a synthetic data set
X, y = make_regression(n_samples = 1000, random_state = 0)
lr = LinearRegression()

result = cross_validate(lr, X, y)
result['test_score'] # r2 score
```

```python
# Automatic parameter searches

# All estimators have hyper-parameters tha can be tuned
# We've got some tools to find the best parameter combinations via cross-validastion

from sklearn.datasets import make_regression
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import RandomizedSearchCV, train_test_split
from scipy.stats import randint

# create a synthetic dataset

X,y = make_regression(n_samples = 20640,
                      n_features = 8,
                      noise = 0.1,
                      random_state = 0)

X_train, X_test, y_train, y_test = train_test_split(X,y,random_state = 0)

# Define the parameter space that will be searched

param_distributions = {'n_estimators': randint(1,5),
                       'max_depth': randint(5,10)}

# Create a searchCV object and fit it to the data

search = RandomizedSearchCV(
    estimator = RandomForestRegressor(random_state = 0),
    n_iter = 5,
    param_distributions = param_distributions,
    random_state = 0)

search.fit(X_train, y_train)
```

```python
search.fit(X_train, y_train)
```

```python
search.best_params_
```

```python
# the search object now acts like a normal random forest estimator
# with max_depth=9 and n_estimators=4
search.score(X_test, y_test)
```
